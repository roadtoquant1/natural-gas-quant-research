# 03 · Model review

Distinguish four things:
* **Fitting:** estimating coefficients.
* **Inference:** how sure we are about them (standard errors, p-values), in-sample.
* **Prediction:** performance on data the model never saw.
* **Backtesting:** turning predictions into simulated trades with costs.

Test-period sections appear only after the final phase has run.

```python
import sys, json
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
import pandas as pd
from IPython.display import Image, Markdown, display
pd.set_option("display.max_columns", 40); pd.set_option("display.width", 200)
TABLES, FIGS, PROC = ROOT / "reports/tables", ROOT / "reports/figures", ROOT / "data/processed"
def show_fig(name):
    p = FIGS / name
    display(Image(str(p)) if p.exists() else Markdown(f"_{name} not generated yet_"))
def table(name):
    p = TABLES / f"{name}.csv"
    return pd.read_csv(p) if p.exists() else Markdown(f"_{name} not generated yet_")
```
